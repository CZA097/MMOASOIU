# Задача №25. Удаление выбросов по IQR
# ---------------------------------------------------

```python
# выберите один числовой столбец
num_col = 'current_price'    # <- замените на имя вашего столбца
```

```python
Q1 = df[num_col].quantile(0.25)
Q3 = df[num_col].quantile(0.75)
IQR = Q3 - Q1
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
outliers = df[(df[num_col] < lower_bound) | (df[num_col] > upper_bound)]
print("Количество выбросов:", outliers.shape[0])
print(outliers[[num_col]].head())
```

```python
df_clean = df[(df[num_col] >= lower_bound) & (df[num_col] <= upper_bound)].copy()
```

```python
print("Размер до удаления выбросов:", df.shape)
print("Размер после удаления выбросов:", df_clean.shape)
```
